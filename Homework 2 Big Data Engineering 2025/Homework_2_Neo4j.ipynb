## 2. Preprocessing Functions

Some columns have pseudo-structured data, like dictionaries and lists. In particular:

**Lists**

*   Conditions
*   Investigators/Contacts
*   Sponsors/Collaborators
*   City of Sponsor/Collaborator
*   State of Sponsor/Collaborator
*   Country of Sponsor/Collaborator
*   Fields of Research (ANZSRC 2020)
*   Collaborating Funders
*   Funder Group
*   Funder Country
*   RCDC Categories
*   HRCS HC Categories
*   HRCS RAC Categories
*   Cancer Types
*   CSO Categories

**List of two elements**

*   Age (min ; max)

For better analytics, we have changed manually the type of these columns with a preprocessing script.

```python
def split_field(value, delimiter=";"):
  """
  Splits a string by a given delimiter into a list of stripped strings.
  Returns an empty list if the input value is None.
  """
  if not value:
      return []

  return [item.strip() for item in value.split(delimiter)]

def parse_age(age_string):
  """
  Parses an age string (e.g., "10 years - 12 months") into a dictionary
  containing 'min_years' and 'max_years' (both in years).
  Handles 'None' or 'n/a' inputs gracefully.
  """
  if not age_string or age_string.strip().lower() == "n/a - n/a":
      return [None, None]

  min_age, max_age = None, None
  parts = [p.strip() for p in age_string.split('-')]

  def convert_to_years(value_part_str):
    if "n/a" in value_part_str.lower():
        return None

    num_match = re.search(r'(\d+\.?\d*)\s*(year|years|month|months|week|weeks|day|days|hour|hours)', value_part_str, re.IGNORECASE)
    if not num_match:
        return None

    num = float(num_match.group(1))
    unit = num_match.group(2).lower()

    if "year" in unit:
        return num
    elif "month" in unit:
        return num / 12
    elif "week" in unit:
        return num / 52
    elif "day" in unit:
        return num / 365
    elif "hour" in unit:
        return num / (365 * 24)
    return None

  if len(parts) > 0:
      min_age = convert_to_years(parts[0])
  if len(parts) > 1:
      max_age = convert_to_years(parts[1])

  return [min_age, max_age]

def to_int_or_null(value):
    try:
        return int(float(value))
    except (ValueError, TypeError):
        return None

def to_float_or_null(value):
    try:
        return float(value)
    except (ValueError, TypeError):
        return None
```

## 3. Dataset Load

Before loading data, we define the Cypher query `LOAD_TRIALS_QUERY`, which acts as our blueprint for structuring the dataset in Neo4j AuraDB.

This query processes clinical trial data in batches, using `UNWIND` to handle multiple records at once. It first creates or updates a `ClinicalTrial` node for each entry, setting its core properties. Then, it iteratively creates or merges related nodes (e.g., `Condition`, `SponsorCollaborator`, `Country`, `Funder`) and establishes specific relationships between them and the `ClinicalTrial` node (like [:HAS_CONDITION], [:FUNDED_BY]). The query includes logic to clean and standardize input data (e.g., trimming, lowercasing) and uses `WITH` clauses to manage context flow, ensuring all parts of the data are correctly linked within the graph.

```python
LOAD_TRIALS_QUERY = """
UNWIND $trials_batch AS row
MERGE (ct:ClinicalTrial {TrialID: row.trial_id})
ON CREATE SET
  ct.Title = row.title,
  ct.BriefTitle = row.brief_title,
  ct.Acronym = row.acronym,
  ct.Abstract = row.abstract,
  ct.StartDate = CASE WHEN row.start_date <> '' THEN date(row.start_date) ELSE null END,
  ct.StartYear = row.start_year,
  ct.EndDate = CASE WHEN row.end_date <> '' THEN date(row.end_date) ELSE null END,
  ct.CompletionYear = row.completion_year,
  ct.Phase = row.phase,
  ct.StudyType = row.study_type,
  ct.RecruitmentStatus = row.recruitment_status,
  ct.NumberOfParticipants = row.number_of_participants,
  ct.Intervention = row.intervention_desc,
  ct.Gender = row.gender,
  ct.Registry = row.registry,
  ct.SourceLinkout = row.source_linkout,
  ct.AltmetricAttentionScore = row.altmetric_score,
  ct.DimensionsURL = row.dimensions_url,
  ct.minAgeYears = row.min_age_years,
  ct.maxAgeYears = row.max_age_years
WITH ct, row

// Condition management
WHERE row.conditions IS NOT NULL AND size(row.conditions) > 0
UNWIND row.conditions AS rawConditionName
WITH ct, row, rawConditionName
WHERE rawConditionName IS NOT NULL AND trim(rawConditionName) <> ""
MERGE (c:Condition {name: trim(lower(rawConditionName))})
MERGE (ct)-[:HAS_CONDITION]->(c)
WITH DISTINCT ct, row

// Sponsor/Collaborators management
WHERE row.sponsors IS NOT NULL AND size(row.sponsors) > 0
UNWIND range(0, size(row.sponsors) - 1) as i
WITH ct, row, i, row.sponsors[i] as sponsorName
WHERE sponsorName IS NOT NULL AND trim(sponsorName) <> ""
MERGE (sc:SponsorCollaborator {name: trim(sponsorName)})
MERGE (ct)-[:HAS_SPONSOR_COLLABORATOR]->(sc)

// City management per sponsor
WITH ct, row, i, sc // <-- CORREZIONE: Passa il contesto
WHERE i < size(coalesce(row.sponsor_cities, [])) AND row.sponsor_cities[i] IS NOT NULL AND trim(row.sponsor_cities[i]) <> ""
MERGE (city:City {name: trim(row.sponsor_cities[i])})
MERGE (sc)-[:LOCATED_IN_CITY]->(city)

// State management per sponsor
WITH ct, row, i, sc // <-- CORREZIONE: Passa il contesto
WHERE i < size(coalesce(row.sponsor_states, [])) AND row.sponsor_states[i] IS NOT NULL AND trim(row.sponsor_states[i]) <> ""
MERGE (st:State {name: trim(row.sponsor_states[i])})
MERGE (sc)-[:LOCATED_IN_STATE]->(st)

// Country management per sponsor
WITH ct, row, i, sc // <-- CORREZIONE: Passa il contesto
WHERE i < size(coalesce(row.sponsor_countries, [])) AND row.sponsor_countries[i] IS NOT NULL AND trim(row.sponsor_countries[i]) <> ""
MERGE (country:Country {name: trim(lower(row.sponsor_countries[i]))})
MERGE (sc)-[:IN_COUNTRY]->(country)

// Aggrega dopo aver processato tutte le località per l'indice 'i' corrente
WITH DISTINCT ct, row // Questa riga raggruppa i risultati per (ct, row) prima di passare al blocco successivo

// Investigators/Contacts management
WHERE row.investigators IS NOT NULL AND size(row.investigators) > 0
UNWIND row.investigators AS rawInvestigatorName
WITH ct, row, rawInvestigatorName
WHERE rawInvestigatorName IS NOT NULL AND trim(rawInvestigatorName) <> ""
MERGE (inv:InvestigatorContact {name: trim(lower(rawInvestigatorName))})
MERGE (ct)-[:HAS_INVESTIGATOR_CONTACT]->(inv)
WITH DISTINCT ct, row

// Collaborating Funders management
WHERE row.collaborating_funders IS NOT NULL AND size(row.collaborating_funders) > 0
UNWIND row.collaborating_funders AS rawFunderName
WITH ct, row, rawFunderName
WHERE rawFunderName IS NOT NULL AND trim(rawFunderName) <> ""
MERGE (f:Funder {name: trim(lower(rawFunderName))})
MERGE (ct)-[:FUNDED_BY]->(f)

// Funder Countries management
WITH ct, row, f // Passa il Funder 'f' creato/matchato sopra
WHERE row.funder_countries IS NOT NULL AND size(row.funder_countries) > 0
UNWIND row.funder_countries AS rawFunderCountryName
WITH ct, row, f, rawFunderCountryName
WHERE rawFunderCountryName IS NOT NULL AND trim(rawFunderCountryName) <> ""
MERGE (fc:Country {name: trim(lower(rawFunderCountryName))})
MERGE (f)-[:IN_COUNTRY]->(fc)
WITH DISTINCT ct, row

// Funder Group management
WHERE row.funder_groups IS NOT NULL AND size(row.funder_groups) > 0
UNWIND row.funder_groups AS rawFunderGroupName
WITH ct, row, rawFunderGroupName
WHERE rawFunderGroupName IS NOT NULL AND trim(rawFunderGroupName) <> ""
MERGE (fg:FunderGroup {name: trim(lower(rawFunderGroupName))})
MERGE (ct)-[:BELONGS_TO_FUNDER_GROUP]->(fg)
WITH DISTINCT ct, row

// Fields of Research
WHERE row.fields_of_research IS NOT NULL AND size(row.fields_of_research) > 0
UNWIND row.fields_of_research AS rawFieldName
WITH ct, row, rawFieldName
WHERE rawFieldName IS NOT NULL AND trim(rawFieldName) <> ""
MERGE (forNode:FieldOfResearch {name: trim(rawFieldName)})
MERGE (ct)-[:HAS_FIELD_OF_RESEARCH]->(forNode)
WITH DISTINCT ct, row

// RCDC Categories
WHERE row.rcdc_categories IS NOT NULL AND size(row.rcdc_categories) > 0
UNWIND row.rcdc_categories AS rawCategoryName
WITH ct, row, rawCategoryName WHERE rawCategoryName IS NOT NULL AND trim(rawCategoryName) <> ""
MERGE (rcdc:RCDCCategory {name: trim(rawCategoryName)})
MERGE (ct)-[:HAS_RCDC_CATEGORY]->(rcdc)
WITH DISTINCT ct, row

// HRCS HC Categories
WHERE row.hrcs_hc_categories IS NOT NULL AND size(row.hrcs_hc_categories) > 0
UNWIND row.hrcs_hc_categories AS rawCategoryName
WITH ct, row, rawCategoryName WHERE rawCategoryName IS NOT NULL AND trim(rawCategoryName) <> ""
MERGE (hrcshc:HRCSHCCategory {name: trim(rawCategoryName)})
MERGE (ct)-[:HAS_HRCS_HC_CATEGORY]->(hrcshc)
WITH DISTINCT ct, row

// HRCS RAC Categories
WHERE row.hrcs_rac_categories IS NOT NULL AND size(row.hrcs_rac_categories) > 0
UNWIND row.hrcs_rac_categories AS rawCategoryName
WITH ct, row, rawCategoryName WHERE rawCategoryName IS NOT NULL AND trim(rawCategoryName) <> ""
MERGE (hrcsrac:HRCSRACCategory {name: trim(rawCategoryName)})
MERGE (ct)-[:HAS_HRCS_RAC_CATEGORY]->(hrcsrac)
WITH DISTINCT ct, row

// Cancer Types
WHERE row.cancer_types IS NOT NULL AND size(row.cancer_types) > 0
UNWIND row.cancer_types AS rawCancerName
WITH ct, row, rawCancerName WHERE rawCancerName IS NOT NULL AND trim(rawCancerName) <> ""
MERGE (cancer:CancerType {name: trim(rawCancerName)})
MERGE (ct)-[:STUDIES_CANCER_TYPE]->(cancer)
WITH DISTINCT ct, row

// CSO Categories
WHERE row.cso_categories IS NOT NULL AND size(row.cso_categories) > 0
UNWIND row.cso_categories AS rawCategoryName
WITH ct, row, rawCategoryName WHERE rawCategoryName IS NOT NULL AND trim(rawCategoryName) <> ""
MERGE (cso:CSOCategory {name: trim(rawCategoryName)})
MERGE (ct)-[:HAS_CSO_CATEGORY]->(cso)
WITH DISTINCT ct, row

// AHC management
WHERE row.ahc IS NOT NULL AND size(row.ahc) > 0
UNWIND row.ahc AS rawAhcName
WITH ct, row, rawAhcName WHERE rawAhcName IS NOT NULL AND trim(rawAhcName) <> ""
MERGE (ahcNode:AcademicHealthCenter {name: trim(rawAhcName)})
MERGE (ct)-[:ASSOCIATED_WITH_AHC]->(ahcNode)
WITH DISTINCT ct

RETURN count(ct) AS loaded_trials_in_batch
"""
```

This function loads CSV data in batches into Neo4j. It preprocesses each row, then uses the `LOAD_TRIALS_QUERY` to create or update nodes and relationships in the graph, optimizing for bulk import performance.

```python
def load_data_to_neo4j(driver):
  batch_size = 10
  trials_batch = []
  total_trials_processed_in_batches = 0
  total_csv_rows_read = 0

  print(f"Inizio caricamento con batch_size: {batch_size}")

  with open(csvPath, 'r', encoding='utf-8') as csvfile:
    reader = csv.DictReader(csvfile)
    column_names = reader.fieldnames

    print(f"Nomi colonne rilevati dal CSV: {column_names}")

    for i, row_data in enumerate(reader):
      total_csv_rows_read += 1
      try:
        age_min_max = parse_age(row_data.get("Age", ""))

        trial_params = {
            "trial_id": row_data.get("Trial ID"),
            "title": row_data.get("Title"),
            "brief_title": row_data.get("Brief title"),
            "acronym": row_data.get("Acronym"),
            "abstract": row_data.get("Abstract"),
            "start_date": row_data.get("Start date"),
            "start_year": to_int_or_null(row_data.get("Start Year")),
            "end_date": row_data.get("End Date"),
            "completion_year": to_int_or_null(row_data.get("Completion Year")),
            "phase": row_data.get("Phase"),
            "study_type": row_data.get("Study Type"),
            "conditions": split_field(row_data.get("Conditions")),
            "recruitment_status": row_data.get("Recruitment Status"),
            "number_of_participants": to_int_or_null(row_data.get("Number of Participants")),
            "intervention_desc": row_data.get("Intervention"),
            "gender": row_data.get("Gender"),
            "min_age_years": age_min_max[0],
            "max_age_years": age_min_max[1],
            "registry": row_data.get("Registry"),
            "investigators": split_field(row_data.get("Investigators/Contacts")),
            "sponsors": split_field(row_data.get("Sponsors/Collaborators")),
            "sponsor_cities": split_field(row_data.get("City of Sponsor/Collaborator")),
            "sponsor_states": split_field(row_data.get("State of Sponsor/Collaborator")),
            "sponsor_countries": split_field(row_data.get("Country of Sponsor/Collaborator")),
            "collaborating_funders": split_field(row_data.get("Collaborating Funders")),
            "funder_groups": split_field(row_data.get("Funder Group")),
            "funder_countries": split_field(row_data.get("Funder Country")),
            "source_linkout": row_data.get("Source Linkout"),
            "altmetric_score": to_float_or_null(row_data.get("Altmetric Attention Score")),
            "dimensions_url": row_data.get("Dimensions URL"),
            "fields_of_research": split_field(row_data.get("Fields of Research (ANZSRC 2020)")),
            "rcdc_categories": split_field(row_data.get("RCDC Categories")),
            "hrcs_hc_categories": split_field(row_data.get("HRCS HC Categories")),
            "hrcs_rac_categories": split_field(row_data.get("HRCS RAC Categories")),
            "cancer_types": split_field(row_data.get("Cancer Types")),
            "cso_categories": split_field(row_data.get("CSO Categories")),
            "ahc": split_field(row_data.get("AHC")) if row_data.get("AHC") else []
        }
        trials_batch.append(trial_params)

        if len(trials_batch) >= batch_size:
          with driver.session(database="neo4j") as session:
            result = session.run(LOAD_TRIALS_QUERY, trials_batch=trials_batch)
            summary = result.consume()
            batch_count = len(trials_batch)
            total_trials_processed_in_batches += batch_count
            print(f"Caricato batch di {batch_count} trial. Totale finora: ~{total_csv_rows_read}")
          trials_batch = []

      except Exception as e:
        print(f"Errore alla riga {i+1} del CSV ({row_data.get('Trial ID', 'ID SCONOSCIUTO')}): {e}")
        print(f"Dati riga problematica: {row_data}")
        continue

    if trials_batch:
      with driver.session(database="neo4j") as session:
        result = session.run(LOAD_TRIALS_QUERY, trials_batch=trials_batch)
        summary = result.consume()
        batch_count = len(trials_batch)
        total_trials_processed_in_batches += batch_count
        print(f"Caricato ultimo batch di {batch_count} trial. Nodi creati: {summary.counters.nodes_created}, Relazioni create: {summary.counters.relationships_created}")

    print("Caricamento dati completato.")

def get_results_analytics(driver, cypher_query):

  df = pd.DataFrame()
  error = False
  try:
    with driver.session(database = 'neo4j') as session:
      print(f"\nEsecuzionde della query:\n {cypher_query}")
      results = session.run(cypher_query)
      df = pd.DataFrame(record.data() for record in results)
      print(f"Risultati ottenuti:\n {df}")
  except Exception as e:
      print(f"Errore durante l'esecuzione della query: {e}")
      error = True
  return df, error

def get_graph_analytics(driver, cypher_query):
  graph = None
  error = False
  try:
    with driver.session(database = 'neo4j') as session:
      print(f"\nEsecuzionde della query:\n {cypher_query}")
      results = session.run(cypher_query)
      graph = results.graph()
  except Exception as e:
      print(f"Errore durante l'esecuzione della query: {e}")
      error = True
  return graph, error
```

```python
def save_csv(df, number_of):
  if not os.path.exists(result_path):
    os.makedirs(result_path)
  df.to_csv(f"{result_path}/results_neo4j_n{number_of}.csv", index=False, encoding='utf-8')
  print(f"Risultati query n{number_of} salvati in csv")

def save_graph(graph, number_of):
    if graph and (graph.nodes or graph.relationships): # Controlla se il grafo non è vuoto
      net = Network(notebook=False, height="750px", width="100%", bgcolor="#222222", font_color="white", cdn_resources='remote')

      # Aggiungi nodi al grafo Pyvis
      for node in graph.nodes:
          node_label = node.get("id") or node.get("nct_id") or list(node.labels)[0]
          net.add_node(str(node.id), label=node_label,
                        title=f"ID: {node.id}\nLabels: {', '.join(node.labels)}\nProperties: {dict(node)}",
                        color="#ADD8E6") # Colore predefinito

      # Aggiungi relazioni al grafo Pyvis (se la tua query le restituisce)
      for rel in graph.relationships:
          net.add_edge(str(rel.start_node.id), str(rel.end_node.id),
                        title=rel.type, # Tipo di relazione al passaggio del mouse
                        label=rel.type, # Testo sulla relazione
                        color="#B0C4DE") # Colore delle relazioni

      net.set_options("""
      var options = {
        "physics": {
          "forceAtlas2Based": {
            "gravitationalConstant": -500,
            "centralGravity": 0.01,
            "springLength": 100,
            "springConstant": 0.08,
            "avoidOverlap": 0.5
          },
          "minVelocity": 0.75,
          "solver": "forceAtlas2Based"
        }
      }
      """)

      # Salva il grafo come file HTML
      graph_output_html_path = os.path.join(result_path, "graphs", f"clinical_trials_graph_n{number_of}.html")
      net.save_graph(graph_output_html_path)
      print(f"Visualizzazione grafo salvata in: {graph_output_html_path}")
      print(f"Apri '{graph_output_html_path}' nel tuo browser per vedere il grafo interattivo.")
    else:
      print("Il grafo è vuoto, nessuna visualizzazione generata.")
```

This function is crucial for optimizing query performance in Neo4j. It iterates through a predefined list of Cypher `CREATE INDEX` statements, applying them to various node labels and properties (e.g., `ClinicalTrial.TrialID`, `Condition.name`). This ensures faster data retrieval and relationship traversal, significantly improving the efficiency of subsequent analytical queries.

```python
def create_indexes(driver):
    queries = [
        "CREATE INDEX clinicaltrial_trialid IF NOT EXISTS FOR (n:ClinicalTrial) ON (n.TrialID);",
        "CREATE INDEX condition_name IF NOT EXISTS FOR (n:Condition) ON (n.name);",
        "CREATE INDEX sponsorcollaborator_name IF NOT EXISTS FOR (n:SponsorCollaborator) ON (n.name);",
        "CREATE INDEX investigatorcontact_name IF NOT EXISTS FOR (n:InvestigatorContact) ON (n.name);",
        "CREATE INDEX funder_name IF NOT EXISTS FOR (n:Funder) ON (n.name);",
        "CREATE INDEX country_name IF NOT EXISTS FOR (n:Country) ON (n.name);",
        "CREATE INDEX city_name IF NOT EXISTS FOR (n:City) ON (n.name);",
        "CREATE INDEX state_name IF NOT EXISTS FOR (n:State) ON (n.name);",
        "CREATE INDEX cancertype_name IF NOT EXISTS FOR (n:CancerType) ON (n.name);",
        "CREATE INDEX ahc_name IF NOT EXISTS FOR (n:AcademicHealthCenter) ON (n.name);"
    ]
    with driver.session(database="neo4j") as session:
        for query in queries:
            try:
                session.run(query)
                print(f"Eseguito: {query}")
            except Exception as e:
                print(f"Errore durante la creazione dell'indice '{query}': {e}")
        print("Creazione indici completata.")
```

3. Top 10 Most Frequent Medical Conditions

```python
# Top 10 most frequent medical conditions
query_n3 = """
MATCH (ct: ClinicalTrial)-[r: HAS_CONDITION]-(c: Condition)
WHERE c.name IS NOT NULL
RETURN c.name AS `Condition Name`, count(r) AS `Count per Condition`
ORDER BY `Count per Condition` DESC
LIMIT 10
"""

query_graph_n3 = """
MATCH (ct:ClinicalTrial)-[r:HAS_CONDITION]-(c:Condition)
WHERE c.name IS NOT NULL
WITH c, count(r) AS relCount, collect({ct: ct, r: r}) AS relationshipsAndTrials
ORDER BY relCount DESC
UNWIND relationshipsAndTrials AS item
LIMIT 400
RETURN item.ct, item.r, c
"""

df_n3, error = get_results_analytics(driver, query_n3)

if error:
  print("Ripetere la query n3, df vuoto")
else:
  save_csv(df_n3, 3)

graph_n3, error = get_graph_analytics(driver, query_graph_n3)

if error:
  print("Ripetere la query n3, graph vuoto")
else:
  save_graph(graph_n3, 3)
```

4. Country with the Highest Average Participants per Study Type

```python
# Countries with the highest average number of participants per study
query_n4 = """
MATCH (ct: ClinicalTrial)-[cs: HAS_SPONSOR_COLLABORATOR]-(s: SponsorCollaborator)-[sc: IN_COUNTRY]-(co: Country)
WHERE ct.NumberOfParticipants IS NOT NULL
AND co.name IS NOT NULL
AND ct.StudyType IS NOT NULL
WITH ct.StudyType AS studyType, co.name AS Country, avg(ct.NumberOfParticipants) AS AvgParticipants

ORDER BY ct.StudyType ASC, AvgParticipants DESC

WITH studyType, COLLECT({country: Country, average: AvgParticipants}) AS countryAvgs
UNWIND countryAvgs[0..1] AS topCountryData

RETURN studyType AS `Study Type`,
topCountryData.country AS `Country`,
topCountryData.average AS `AvgParticipants per StudyType`
ORDER BY `AvgParticipants per StudyType` DESC
LIMIT 20
"""

df_n4, error = get_results_analytics(driver, query_n4)

if error:
  print("Ripetere la query n4, df vuoto")
else:
  save_csv(df_n4, 4)

```

5. Cities that Most Frequently Treated a Specific Condition

```python
# Città che hanno trattato maggiormente una determinata condizione (Per condizione)

query_n5 = """
MATCH (con: Condition)-[hc: HAS_CONDITION]-(ct: ClinicalTrial)-[cs: HAS_SPONSOR_COLLABORATOR]-(s: SponsorCollaborator)-[li: LOCATED_IN_CITY]-(ci: City)
WHERE s.name IS NOT NULL
AND ci.name IS NOT NULL
AND con.name IS NOT NULL

LIMIT 90000

WITH con.name AS Condition, ci.name AS City, ct
WITH Condition, City, count(DISTINCT ct) AS `Count per Condition-City`

ORDER BY `Count per Condition-City` DESC

WITH Condition, COLLECT({city: City, count: `Count per Condition-City`}) AS cityCounts
UNWIND cityCounts[0..1] AS topCityData

RETURN Condition AS `Condition`,
topCityData.city AS `City`,
topCityData.count AS `Count per Condition-City`
ORDER BY `Count per Condition-City` DESC
LIMIT 20
"""

query_graph_n5 = """
MATCH (con: Condition)-[hc: HAS_CONDITION]-(ct: ClinicalTrial)-[cs: HAS_SPONSOR_COLLABORATOR]-(s: SponsorCollaborator)-[li: LOCATED_IN_CITY]-(ci: City)
WHERE s.name IS NOT NULL
AND ci.name IS NOT NULL
AND con.name IS NOT NULL

LIMIT 90000

WITH con.name AS Condition, ci.name AS City, ct
WITH Condition, City, count(DISTINCT ct) AS `Count per Condition-City`

ORDER BY `Count per Condition-City` DESC

WITH Condition, COLLECT({city: City, count: `Count per Condition-City`}) AS cityCounts
UNWIND cityCounts[0..1] AS topCityData

MATCH (fcon: Condition {name: Condition})
-[fhc: HAS_CONDITION]-(fct: ClinicalTrial)
-[fcs: HAS_SPONSOR_COLLABORATOR]-(fs: SponsorCollaborator)
-[fli: LOCATED_IN_CITY]-(fci: City {name: topCityData.city})

RETURN fcon, fhc, fct, fcs, fs, fli, fci
LIMIT 200
"""

df_n5, error = get_results_analytics(driver, query_n5)

if error:
  print("Ripetere la query n5, df vuoto")
else:
  save_csv(df_n5, 5)

graph_n5, error = get_graph_analytics(driver, query_graph_n5)

if error:
  print("Ripetere la query n5, graph vuoto")
else:
  save_graph(graph_n5, 5)
```
